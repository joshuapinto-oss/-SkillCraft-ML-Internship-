```python
#  SKILLCRAFT TECHNOLOGY – TASK 02
#  K-Means Clustering – Customer Segmentation
#  Joshua Pinto
#  Dataset : Mall_Customers.csv

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Load data
df = pd.read_csv(r'C:\Users\hp\Desktop\Internship\task 2\Mall_Customers.csv')

# Select columns
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]

# Apply KMeans
kmeans = KMeans(n_clusters=5, random_state=42)
df['Cluster'] = kmeans.fit_predict(X)

# Plot clusters
colors = ['red', 'blue', 'green', 'orange', 'purple']

for i in range(5):
    points = df[df['Cluster'] == i]
    plt.scatter(points['Annual Income (k$)'], points['Spending Score (1-100)'],
                color=colors[i], label=f'Cluster {i+1}')

# Plot center points
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], color='black', marker='X', s=200, label='Centers')

plt.title('Customer Groups')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score')
plt.legend()
plt.show()

# Show result
print(df.groupby('Cluster')[['Annual Income (k$)', 'Spending Score (1-100)']].mean().round(1))
```

```python
# Custom Value Prediction
my_income = 80    
my_spending = 85     

my_data = [[my_income, my_spending]]
result = kmeans.predict(my_data)
my_cluster = result[0]

print(f"Income: {my_income}k$  |  Spending Score: {my_spending}")
print(f"This customer belongs to --> Cluster {my_cluster + 1}")

# Plot clusters
colors = ['red', 'blue', 'green', 'orange', 'purple']

for i in range(5):
    points = df[df['Cluster'] == i]
    plt.scatter(points['Annual Income (k$)'], points['Spending Score (1-100)'],
                color=colors[i], label=f'Cluster {i+1}', alpha=0.6)

# Plot centers
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], color='black', marker='X', s=200, label='Centers')

# ⭐ Plot your custom point
plt.scatter(my_income, my_spending, color=colors[my_cluster],
            marker='*', s=500, edgecolors='black', linewidths=1.5,
            label=f'My Customer (Cluster {my_cluster + 1})')

plt.annotate(f'  ← My Customer\n     Income={my_income}k$\n     Spending={my_spending}',
             xy=(my_income, my_spending),
             fontsize=9, color='black',
             bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.7))

plt.title('Customer Groups with Custom Prediction')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score')
plt.legend()
plt.tight_layout()
plt.show()
```
